# Cell 2: Importing libraries

```python
import re
import spacy
from collections import Counter
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

#load spacy english model that is used for name entity recognition

nlp = spacy.load("en_core_web_sm")
print("All imports were imported successfully !")
```

# Cell 3: Load the text file
<h1 style='font-size: 15px; font-weight: bold;'> The text file is about: the role of flowers in human language and culture over time, and how they are useful for natural language processing (NLP).</h1>

```python
# we are going to load our text file, ours is named flowers.txt

file_path = "flowers.txt"

with open(file_path, "r", encoding="utf-8") as f:
    raw_text = f.read()

print(f"File loaded: {file_path}")
print(f"Total characters: {len(raw_text)}")
print("\n--- Preview (first 500 characters) ---")
print(raw_text[:500])
```

# Cell 4: Sentence Segmentation

```python

# this function splits text into sentences using nltk punkt mpdels
#it detects sentence boundaries based on punct & cap
sentences = sent_tokenize(raw_text)

print(f"Total sentences found: {len(sentences)}")
print("\n--- First 5 sentences ---")
for i, sent in enumerate(sentences[:5], 1):
    print(f"  [{i}] {sent}")
```

# Cell 5: Word Tokenization

```python
#word_tokenize splits text into individual word tokens

# tokenize + turn the letters into lowercase
all_tokens = word_tokenize(raw_text.lower())

#remove punct/nbrs
words_only = [w for w in all_tokens if w.isalpha()]

print(f"Total tokens (with punct): {len(all_tokens)}")
print(f"Alphabetic word tokens only:  {len(words_only)}")
print(f"\n First 20 tokens:")
print(words_only[:20])


```

# Cell 6: Words & their frequencies (descending order)

```python
# À ajouter dans Cell 5 ou 6 :
word_freq = Counter(words_only)
print("Words with their frequencies :")
for word, count in word_freq.most_common(20):
    print(f"  {word}: {count}")
```

# Cell 7: Stop-words Removal

<h1 style='font-size: 15px;'>We are going to remove stop-words here using 2 distint methods which are: </h1>
<h1 style=' font-size: 15px; font-weight: bold; color: pink;'>Method 1:</h1><h1 style='font-size: 15px;'>Standard NLTK english stop-words.</h1>
<h1 style=' font-size: 15px; font-weight: bold; color: pink;'>Method 2:</h1><h1 style='font-size: 15px;'>Our own custom stop-words that we added on top of the standard list.</h1>

```python
# method 1: standard stop-words
standard_stopwords = set(stopwords.words('english'))
filtered_standard = [w for w in words_only if w not in standard_stopwords]

print(f"Standard stop-words size: {len(standard_stopwords)}")
print(f"Words after standard stop-words removal: {len(filtered_standard)}")
print("Sample:", filtered_standard[:15])

print()

# method 2: custom stop-words
custom_input = [
    "flowers", "flower", "used", "use", "using",
    "language", "text", "words", "word",
    "meanings", "meaning",
    "time", "times", "century", "centuries",
    "example", "examples",
    "often", "many", "also",
    "make", "makes", "making",
    "such", "like",
    "across", "different",
    "human", "humans",
    "systems", "system",
    "data", "information",
    "present", "based",
    "including", "related",
    "context", "contexts"
]

user_stopwords = set(w.strip().lower() for w in custom_input)

# merge both sets
combined_stopwords = standard_stopwords | user_stopwords

filtered_custom = [w for w in words_only if w not in combined_stopwords]

print(f"Custom stop-words added: {user_stopwords}")
print(f"Words after combined stop-words removal: {len(filtered_custom)}")
print("Sample:", filtered_custom[:15])

# use combined result for next steps
clean_words = filtered_custom

```

# Cell 8: Top N-grams
<h1 style='font-size: 15px'>An n-gram is a sequence of N consecutive words, we are going to use python s zip() to build n-grams from the word list and rank them by frequency.</h1>

```python
N = 10
# Unigrams (N=1)--> single words
unigram_freq = Counter(clean_words).most_common(N)

# Bigrams (N=2)--> Two-word pairs
bigrams = list(zip(clean_words, clean_words[1:]))
bigram_freq = Counter(bigrams).most_common(N)

# trigrams (N=3)--> three-word sequences
trigrams = list(zip(clean_words, clean_words[1:], clean_words[2:]))
trigram_freq = Counter(trigrams).most_common(N)

print(f" Top {N} Unigrams:")
for word, count in unigram_freq:
    print(f" '{word}' --> {count}x")

print(f" Top {N } Bigrams:")
for gram, count in bigram_freq:
    print(f" '{gram[0]}{gram[1]}' --> {count}x ")

print(f" Top {N} Trigrams:")
for gram, count in trigram_freq:
    print(f" '{gram[0]}{gram[1]}{gram[2]}' --> {count}x'")
```

# Cell 9: POS Tagging
<h1 style='font-size: 15px;'> Speech tagigng assigns a grammatical label to each word.</h1>


```python
# keep the punct for better tagging
tokens_for_pos = word_tokenize(raw_text)

# returns a list of (word, TAG) tuples
pos_tags = pos_tag(tokens_for_pos)

print(f"Total tagged tokens: {len(pos_tags)}")
print(f" \n First 40(word --> POS tag)")
for word, tag in pos_tags[:40]:
    print(f" {word:20s} --> {tag}")


# count how many times each POS tag appears in the word text
tag_counts = Counter(tag for _, tag in pos_tags)
print(f" \n POS Tag Frequency(top10)")
for tag, count in tag_counts.most_common(10):
    print(f" {tag}:{count}")
```

# Cell 10: Named Entity Recognition or NER
<h1 style='font-size: 15px;'>NER detects+classifies real world entities mentioned in the text like: PERSON --> person's name, DATE--> date & time periods...etc.</h1>

```python
# process the full text through the spacy pipeline
doc = nlp(raw_text)

print("Named Entities Found:")
print(f"{'Entity Text':<35} {'Label':<12} {'Description'}")
print("-" * 65)

seen = set()

for ent in doc.ents:   # FIX: use ent consistently
    key = (ent.text.strip(), ent.label_)
    
    if key not in seen:
        seen.add(key)
        print(f"{ent.text.strip():<33} {ent.label_:<12} {spacy.explain(ent.label_)}")

print(f"\nTotal unique named entities: {len(seen)}")
```

# Cell 11: Date Extraction with Regular Expressions
<h1 style='font-size: 15px;'>A regular expression regex used to search text, here it covers all common english date formats and which are: </h1>

<h1 style='font-size: 15px; font-weight: bold;'>DD/MM/YYYY             exp: 25/12/2023 <br>
  MM-DD-YYYY              exp: 12-25-2023<br>
  YYYY-MM-DD  (ISO)       exp: 2023-12-25<br>
  DD Month YYYY           exp: 25 December 2023<br>
  Month DD, YYYY          exp: December 25, 2023<br>
  DD Month YY             exp: 25 Dec 23<br>
  Month YYYY              exp: December 2023<br>
  DD.MM.YYYY              exp: 25.12.2023<br>
</h1>

```python


month_names = (
    r"(?:January|February|March|April|May|June|July|August|September|October|November|December|"
    r"Jan|Feb|Mar|Apr|Jun|Jul|Aug|Sep|Oct|Nov|Dec)"
)

date_pattern = re.compile(
    r"\b"
    r"(?:"
        r"\d{4}[-/]\d{1,2}[-/]\d{1,2}"                          # ISO
        r"|"
        r"\d{1,2}[-/.]\d{1,2}[-/.]\d{2,4}"                      # numeric
        r"|"
        r"\d{1,2}\s+" + month_names + r"\s+\d{2,4}"             # 25 December 2023
        r"|"
        + month_names + r"\s+\d{1,2},?\s+\d{2,4}"               # December 25, 2023
        r"|"
        + month_names + r"\s+\d{4}"                             # December 2023
    r")"
    r"\b",
    re.IGNORECASE
)

dates_found = date_pattern.findall(raw_text)

print(f"Dates found in the text: {len(dates_found)}")

if dates_found:
    for d in dates_found:
        print(f" --> {d}")
else:
    print("No dates detected in the text given, try again!")
```

# Cell 12: Word Info 
<h1 style='font-size: 15px;'> (Stem, Lemma, Definition, Example, Synonyms, Antonyms)</h1>

```python
stemmer    = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def get_word_info(word: str):
    """Print full linguistic analysis for a given word."""
    word = word.lower().strip()
    print(f"\n{'='*50}")
    print(f"  Word: '{word}'")
    print(f"{'='*50}")

    # Stem: Porter Stemmer removes suffixes (e.g. 'running' → 'run')
    print(f"  Stem      : {stemmer.stem(word)}")

    # Lemma: returns the base dictionary form
    lemma_n = lemmatizer.lemmatize(word, pos='n')  # as noun
    lemma_v = lemmatizer.lemmatize(word, pos='v')  # as verb
    print(f"  Lemma     : {lemma_n} (noun) / {lemma_v} (verb)")

    # WordNet synsets: each synset = one meaning of the word
    synsets = wordnet.synsets(word)
    if not synsets:
        print("  No WordNet entry found.")
        return

    # Use the first (most common) meaning
    first = synsets[0]

    # Definition
    print(f"  Definition: {first.definition()}")

    # Example sentence
    examples = first.examples()
    print(f"  Example   : {examples[0] if examples else '(none available)'}")

    # Synonyms: collect all lemma names across all synsets
    synonyms = set()
    for syn in synsets:
        for lem in syn.lemmas():
            name = lem.name().replace('_', ' ')
            if name != word:
                synonyms.add(name)
    print(f"  Synonyms  : {', '.join(sorted(synonyms)[:10]) if synonyms else 'None found'}")

    # Antonyms: follow antonym links in WordNet
    antonyms = set()
    for syn in synsets:
        for lem in syn.lemmas():
            for ant in lem.antonyms():
                antonyms.add(ant.name().replace('_', ' '))
    print(f"  Antonyms  : {', '.join(sorted(antonyms)) if antonyms else 'None found'}")


# ─── Test with a few words from the flowers topic ────────────────────────────
# You can change these words to anything you like
test_words = ["flower", "bloom", "beautiful", "grow"]

for w in test_words:
    get_word_info(w)

```
